```python
def count_up_to(n):
    i=1
    while i<=n:
        yield i
        i+=1
       
gen_func = count_up_to(5) 
for number in gen_func:
    # if(number == 2):
    #     gen_func.close()
    print(number)
```

```python
def shout(func):
    def wrapper(*arg):
        return func(*arg).upper()        
    return wrapper

@shout
def greet(name):
    return f"Hello {name}"
    
print(greet("Kundan"))
        
```
